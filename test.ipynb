```python
def print_building_facts(facts):
        """BuildingRegisterRepository.fetch() 결과를 보기 좋게 출력한다.

        facts 구조: {"title": dict, "floors": list, "zones": list}
        """
        title = facts["title"]
        floors = facts["floors"]
        zones = facts["zones"]

        # ── 표제부 ──────────────────────────────────────────────
        print("=" * 50)
        print(" 건축물대장 표제부")
        print("=" * 50)

        sections = [
            ("[ 식별 ]", [
                ("건물명", "building_name"),
                ("관리대장PK", "main_ledger_pk"),
                ("지번주소", "plat_address"),
                ("도로명주소", "road_address"),
            ]),
            ("[ 규모 ]", [
                ("대지면적", "land_area_sqm", "㎡"),
                ("건축면적", "arch_area_sqm", "㎡"),
                ("연면적", "total_floor_area_sqm", "㎡"),
                ("건폐율", "building_coverage_ratio", "%"),
                ("용적률", "floor_area_ratio", "%"),
            ]),
            ("[ 구조·층수 ]", [
                ("구조", "structure"),
                ("지상층수", "floors_above", "층"),
                ("지하층수", "floors_below", "층"),
                ("높이", "height_m", "m"),
            ]),
            ("[ 용도 ]", [
                ("주용도", "main_usage"),
                ("기타용도", "etc_usage"),
                ("세대수", "household_count"),
                ("가구수", "family_count"),
                ("호수", "ho_count"),
            ]),
            ("[ 시기 ]", [
                ("허가일", "permit_date"),
                ("착공일", "start_date"),
                ("사용승인일", "approval_date"),
            ]),
            ("[ ESG ]", [
                ("에너지효율등급", "energy_grade"),
                ("EPI점수", "energy_epi"),
                ("친환경건축물등급", "green_grade"),
                ("친환경인증점수", "green_cert_score"),
                ("지능형건축물등급", "smart_grade"),
                ("지능형인증점수", "smart_cert_score"),
            ]),
        ]

        for section_title, fields in sections:
            print()
            print(section_title)
            for field in fields:
                label = field[0]
                key = field[1]
                unit = field[2] if len(field) > 2 else ""
                value = title.get(key)
                if value is None:
                    shown = "-"
                else:
                    shown = f"{value}{unit}"
                print(f"  {label:<16} {shown}")

        # ── 층별개요 ────────────────────────────────────────────
        print()
        print("=" * 50)
        print(f" 층별개요  ({len(floors)}개 층)")
        print("=" * 50)
        if not floors:
            print("  (데이터 없음)")
        else:
            print(f"  {'층':<8}{'용도':<22}{'면적':>12}")
            print(f"  {'-' * 42}")
            for floor in floors:
                floor_no = floor.get("floor_no") or "-"
                usage = floor.get("main_usage") or "-"
                area = floor.get("area_sqm")
                area_str = f"{area}㎡" if area is not None else "-"
                print(f"  {floor_no:<8}{usage:<22}{area_str:>12}")

        # ── 지역지구구역 ────────────────────────────────────────
        print()
        print("=" * 50)
        print(f" 지역지구구역  ({len(zones)}개)")
        print("=" * 50)
        if not zones:
            print("  (데이터 없음)")
        else:
            for zone in zones:
                name = zone.get("zone_name") or "-"
                gb = zone.get("zone_gb") or ""
                rep = "  ★대표" if zone.get("is_representative") else ""
                print(f"  {name}  ({gb}){rep}")
        print()
```

```python
print_building_facts(건축허브조회객체.fetch(입력주소4의객체))
```
